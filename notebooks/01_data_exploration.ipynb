```python
def inspect_mat_file(path):

    mat = sio.loadmat(path)
    print(f"\nFile: {path}")
    print(f"{'Variable':<15}{'Shape':<20}{'Dtype'}")
    for k, v in mat.items():
        if k.startswith("__"):
            continue
        shape = getattr(v, "shape", None)
        dtype = getattr(v, "dtype", type(v))
        print(f"{k:<15}{str(shape):<20}{dtype}")
    return mat
```
